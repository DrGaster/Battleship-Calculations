```python
"""
Author:      Glen
Date:        09/24/2024
Description: [String Formatting]
"""
# Create variables name, age, and city with appropriate values (does not have to be specific).
# Use the old-style % formatting to create and print a sentence: "My name is [name], I am [age] years old, and I live in [city]."
# Use the str.format() method to create and print the same sentence. (either way can be done)
# Use f-strings (formatted string literals) to create and print the same sentence. (either way can be used)

name = input("Enter your Name: ")
age = int(input("Enter your age: "))
city = str(input("Enter your City: "))

print("My name is %s, I am %d years old, and I live in %s." % (name, age, city))
print("My name is {}, I am {} years old, and I live in {}.".format(name, age, city))
print(f"My name is {name}, I am {age} years old, and I live in {city}.")
```

```python
"""
Author:      Glen Nguyen
Date:        09/25/2024
Description: [Formatting Tables]
"""

# Takes the User's input for the Number, Title, Origin, and Aspects of the first Hour
Hour_One_Number = int(input("Enter the Number of Hour 1: ")) # Takes the User's input for the Number of the first Hour
Hour_One_Title = input("Enter a Title of Hour 1: ") # Takes the User's input for the Title of the first Hour
Hour_One_Origin = input("Enter the Origin of Hour 1: ") # Takes the User's input for the Origin of the first Hour
Hour_One_Aspects = input("Enter the Aspects of Hour 1: ") # Takes the User's input for the Aspects of the first Hour

# Takes the User's input for the Number, Title, Origin, and Aspects of the second Hour
Hour_Two_Number = int(input("Enter the Number of Hour 2: "))
Hour_Two_Title = input("Enter a Title of Hour 2: ")
Hour_Two_Origin = input("Enter the Origin of Hour 2: ")
Hour_Two_Aspects = input("Enter the Aspects of Hour 2: ")

# Takes the User's input for the Number, Title, Origin, and Aspects of the third Hour
Hour_Three_Number = int(input("Enter the Number of Hour 3: "))
Hour_Three_Title = input("Enter a Title of Hour 3: ")
Hour_Three_Origin = input("Enter the Origin of Hour 3: ")
Hour_Three_Aspects = input("Enter the Aspects of Hour 3: ")

# Creates the column headers for the table
Hour_Number = "Number"
Hour_Title = "Hour"
Hour_Origin = "Origin"
Hour_Aspects = "Aspects"

# Creates a formatted table with the User's input
table = """
----------------------------------------------------------------------------
| {:^6} | {:^20} | {:^13} | {:^24} |
| {:<6} | {:<20} | {:<13} | {:<24} |
| {:<6} | {:<20} | {:<13} | {:<24} |
| {:<6} | {:<20} | {:<13} | {:<24} |
----------------------------------------------------------------------------
""".format(Hour_Number, Hour_Title, Hour_Origin, Hour_Aspects,
           Hour_One_Number, Hour_One_Title, Hour_One_Origin, Hour_One_Aspects,
           Hour_Two_Number, Hour_Two_Title, Hour_Two_Origin, Hour_Two_Aspects,
           Hour_Three_Number, Hour_Three_Title, Hour_Three_Origin, Hour_Three_Aspects)

# Prints the formatted table
print(table)
```
